```python
import pandas as pd
from googletrans import Translator

# Load data
df = pd.read_csv('../Data/Dengue_QA.csv')
translator = Translator()

# Translate Malay questions to English
english_questions = []
for q in df['Question (Malay)']:
    try:
        translated = translator.translate(str(q), src='ms', dest='en')
        english_questions.append(translated.text)
    except Exception as e:
        english_questions.append("")
        print(f"Translation error for '{q}': {e}")

# Add the translations as a new column
df['Question (English)'] = english_questions
```

```python
# Translate Malay answers to English
english_answers = []
for q in df['Answer (Malay)']:
    try:
        translated = translator.translate(str(q), src='ms', dest='en')
        english_answers.append(translated.text)
    except Exception as e:
        english_answers.append("")
        print(f"Translation error for '{q}': {e}")

# Add the translations as a new column
df['Answer (English)'] = english_answers
```

```python
# Save to a new CSV
df.to_csv('../Data/Dengue_QA_multilingual.csv', index=False, encoding='utf-8-sig')
print("Translation complete! File saved as Dengue_QA_multilingual.csv")
```
